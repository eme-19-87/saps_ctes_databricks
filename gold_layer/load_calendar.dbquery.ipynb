```sql
TRUNCATE TABLE saps_ctes.gold.dim_calendario;

-- 2. Insertar los datos
INSERT INTO saps_ctes.gold.dim_calendario (
    calendario_key,
    fecha,
    dia,
    mes,
    anio,
    nombre_dia,
    nombre_mes,
    dia_mes,
    mes_anio,
    anio_mes,
    semana_anio,
    trimestre,
    es_fin_semana,
    es_fin_mes,
    es_fin_anio
)
SELECT DISTINCT
    year(fecha) * 10000 +
    month(fecha) * 100 +
    day(fecha)                          AS calendario_key,
    fecha,
    day(fecha)                          AS dia,
    month(fecha)                        AS mes,
    year(fecha)                         AS anio,
    date_format(fecha, 'EEEE')          AS nombre_dia,
    date_format(fecha, 'MMMM')          AS nombre_mes,
    date_format(fecha, 'dd-MM')         AS dia_mes,
    date_format(fecha, 'MM-yyyy')       AS mes_anio,
    date_format(fecha, 'yyyy-MM')       AS anio_mes,
    weekofyear(fecha)                   AS semana_anio,
    quarter(fecha)                      AS trimestre,
    dayofweek(fecha) IN (1,7)           AS es_fin_semana,
    fecha = last_day(fecha)             AS es_fin_mes,
    fecha = last_day(add_months(fecha, 12) - interval 12 months) AS es_fin_anio
FROM (
    SELECT fecha FROM saps_ctes.silver.datosctes_consultas_patologia
    UNION
    SELECT fecha FROM saps_ctes.silver.datosctes_inmunizaciones
) f;
```
