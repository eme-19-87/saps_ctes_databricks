```sql
 TRUNCATE TABLE saps_ctes.gold.fact_inmunizacion;

    INSERT INTO saps_ctes.gold.fact_inmunizacion (
        saps_key, calendario_key, vacuna_key, cantidad_vacuna
    )
    SELECT
        ds.saps_key,
        dc.calendario_key,
        dv.vacuna_key,
        im.vacunas_cantidad
    FROM saps_ctes.silver.datosctes_inmunizaciones im
    LEFT JOIN saps_ctes.gold.dim_saps ds
        ON im.id_saps = ds.id_saps
       AND CURRENT_DATE BETWEEN ds.tiv AND ds.tfv
    LEFT JOIN saps_ctes.gold.dim_vacuna dv
        ON im.vacunas_tipo = dv.tipo_vacuna
    LEFT JOIN saps_ctes.gold.dim_calendario dc
        ON dc.fecha = im.fecha
    WHERE im.id_saps<>-1 AND im.id_saps<>99 AND im.vacunas_cantidad>-1 AND im.vacunas_tipo!='n/a'
    and im.fecha > '1900-01-01';
```
